```python
import os
import time
import requests
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

def crawl_pure_blouson(start_page=1, end_page=5):
    # 1. 저장할 폴더 셋팅 (바탕화면 절대경로 적용)
    # 파이썬에서 윈도우 경로를 쓸 때는 앞에 r을 붙여주는 것이 안전합니다.
    base_dir = r"C:\Users\user\Desktop\졸작 데이터셋" 
    category_name = "blouson_dataset"
    
    # '졸작 데이터셋\blouson_dataset' 형태로 경로 완성
    save_dir = os.path.join(base_dir, category_name)
    os.makedirs(save_dir, exist_ok=True)
    
    # 2. 크롬 브라우저 설정 (자동화 창)
    options = webdriver.ChromeOptions()
    # options.add_argument('--headless') # 크롬 창을 안 띄우려면 주석 해제
    options.add_argument('--no-sandbox')
    options.add_argument('--disable-dev-shm-usage')
    
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
    wait = WebDriverWait(driver, 10)
    
    total_downloaded = 0
    print(f"👕 무신사 순수 블루종 데이터 수집 시작 (페이지 {start_page} ~ {end_page})")
    print(f"📂 저장 위치: {save_dir}")

    # 3. 페이지 순회 시작
    for page in range(start_page, end_page + 1):
        # 무신사 아우터 대분류(002001) 주소 적용
        url = f"https://www.musinsa.com/categories/item/002001?page={page}"
        driver.get(url)
        time.sleep(2) 
        
        print(f"\n--- {page}페이지 스크롤 및 로딩 중 ---")
        
        # 무한 스크롤: 끝까지 내려서 모든 상품 이미지를 화면에 띄움
        last_height = driver.execute_script("return document.body.scrollHeight")
        while True:
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
            time.sleep(1)
            new_height = driver.execute_script("return document.body.scrollHeight")
            if new_height == last_height:
                break
            last_height = new_height

        try:
            # 4. 정확한 명찰(data-mds='Image')로 썸네일만 타겟팅
            product_images = wait.until(EC.presence_of_all_elements_located(
                (By.CSS_SELECTOR, "img[data-mds='Image']")
            ))
            
            for img in product_images:
                # 이미지 태그에서 상품명(alt)과 주소(src) 추출
                product_name = img.get_attribute('alt')
                img_url = img.get_attribute('src')
                
                if product_name and img_url and ("http" in img_url):
                    # 5. 핵심 필터링: 블루종은 통과, MA-1과 항공점퍼는 컷!
                    if ("블루종" in product_name) and ("항공" not in product_name) and ("MA-1" not in product_name.upper()):
                        try:
                            # 이미지 다운로드 요청
                            response = requests.get(img_url, timeout=5)
                            
                            # 파일명 설정 및 저장 경로 결합
                            file_name = os.path.join(save_dir, f"{category_name}_{total_downloaded + 1}.jpg")
                            
                            with open(file_name, "wb") as f:
                                f.write(response.content)
                                
                            total_downloaded += 1
                            # 터미널 창에 어떤 상품이 다운되었는지 깔끔하게 출력
                            print(f"[저장 성공] {product_name}") 
                            
                        except Exception as e:
                            pass
                    else:
                        pass # 필터링에 걸린 상품은 조용히 패스

            print(f"✅ {page}페이지 완료. (현재 누적: {total_downloaded}장)")
            
        except Exception as e:
            print(f"⚠️ {page}페이지 요소 찾기 실패, 다음 페이지로 넘어갑니다.")
            continue

    print(f"\n🎉 [크롤링 최종 완료] 총 {total_downloaded}장의 순도 100% 블루종 이미지가 저장되었습니다!")
    driver.quit()

# 실행부: 1페이지부터 3페이지만 가볍게 테스트!
if __name__ == "__main__":
    crawl_pure_blouson(start_page=1, end_page=3)
```
