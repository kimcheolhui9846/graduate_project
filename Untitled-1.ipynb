```python
import os
import time
import requests
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait

# ==========================================
# 1. 부모 클래스: BaseCrawler (공통 도구 상자)
# ==========================================
class BaseCrawler:
    def __init__(self, base_dir):
        self.base_dir = base_dir
        
        # [중복 방지] 다운로드 기록 파일 및 메모리 세팅
        self.history_file = os.path.join(self.base_dir, "download_history.txt")
        self.downloaded_urls = set() 
        
        if os.path.exists(self.history_file):
            with open(self.history_file, "r", encoding="utf-8") as f:
                for line in f:
                    self.downloaded_urls.add(line.strip())
            print(f"📚 기존 다운로드 기록 {len(self.downloaded_urls)}개를 불러왔습니다.")
        else:
            print("🌱 다운로드 기록이 없습니다. 새로 수집을 시작합니다.")

        # 크롬 브라우저 세팅
        options = webdriver.ChromeOptions()
        # options.add_argument('--headless') # 브라우저 창을 숨기고 싶다면 맨 앞의 #을 지워주세요.
        options.add_argument('--no-sandbox')
        options.add_argument('--disable-dev-shm-usage')
        
        self.driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
        self.wait = WebDriverWait(self.driver, 10)

    def _create_folder(self, label_name):
        save_path = os.path.join(self.base_dir, label_name)
        os.makedirs(save_path, exist_ok=True)
        return save_path

    def _download_image(self, img_url, save_path, file_name):
        """중복 검사 및 다운로드 공통 로직"""
        # 이미 다운받은 URL이면 즉시 'SKIP' 반환
        if img_url in self.downloaded_urls:
            return "SKIP"
            
        try:
            response = requests.get(img_url, timeout=5)
            with open(os.path.join(save_path, file_name), "wb") as f:
                f.write(response.content)
            
            # 다운로드 성공 시 기록장과 메모리에 URL 영구 추가
            with open(self.history_file, "a", encoding="utf-8") as f:
                f.write(img_url + "\n")
            self.downloaded_urls.add(img_url)
            
            return "SUCCESS"
        except Exception as e:
            return "FAIL"

    def close(self):
        self.driver.quit()

# ==========================================
# 2. 자식 클래스: MusinsaCrawler (무신사 전용)
# ==========================================
class MusinsaCrawler(BaseCrawler):
    def __init__(self, base_dir):
        super().__init__(base_dir) 
        
        # 코트 4종 카테고리 통합
        self.category_map = {
            "002008": {"label": "1_coat", "blacklist": ["패딩", "다운", "점퍼", "푸퍼", "무스탕"]}, # 환절기 코트
            "002007": {"label": "1_coat", "blacklist": ["패딩", "다운", "점퍼", "푸퍼", "무스탕"]}, # 겨울 싱글 코트
            "002024": {"label": "1_coat", "blacklist": ["패딩", "다운", "점퍼", "푸퍼", "무스탕"]}, # 겨울 더블 코트
            "002009": {"label": "1_coat", "blacklist": ["패딩", "다운", "점퍼", "푸퍼", "무스탕"]}  # 겨울 기타 코트
        }
        
        self.download_counts = {}

    def run_crawling(self, start_page=1):
        for cat_id, info in self.category_map.items():
            label_name = info["label"]
            blacklist = info["blacklist"]
            save_path = self._create_folder(label_name)
            
            if label_name not in self.download_counts:
                # 기존 파일 개수를 세어서 번호표 세팅
                existing_files = len([f for f in os.listdir(save_path) if os.path.isfile(os.path.join(save_path, f))])
                self.download_counts[label_name] = existing_files
                
            print(f"\n🚀 무신사 카테고리 [{cat_id}] 수집 시작 (폴더: {label_name})")
            
            page = start_page
            empty_page_streak = 0 # 💡 유령 페이지 함정 탈출용 카운터
            
            while True:
                url = f"https://www.musinsa.com/categories/item/{cat_id}?page={page}"
                self.driver.get(url)
                time.sleep(2)
                
                print(f"--- {page}페이지 확인 중 ---")
                
                # 1차 체크: 상품이 아예 없는 페이지 감지
                initial_check = self.driver.find_elements(By.CSS_SELECTOR, "img[data-mds='Image']")
                if not initial_check:
                    print(f"📢 {page}페이지에 상품 태그가 전혀 없습니다! 카테고리 [{cat_id}] 수집을 종료합니다.")
                    break 
                
                # Lazy Loading 대응 스크롤
                for _ in range(15):
                    self.driver.execute_script("window.scrollBy(0, 800);")
                    time.sleep(0.5)
                self.driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
                time.sleep(1)

                try:
                    product_images = self.driver.find_elements(By.CSS_SELECTOR, "img[data-mds='Image']")
                    page_success = 0
                    page_skipped = 0
                    
                    for img in product_images:
                        product_name = img.get_attribute('alt')
                        img_url = img.get_attribute('src')
                        
                        if product_name and img_url and ("http" in img_url):
                            name_lower = product_name.lower()
                            is_excluded = any(bad_word in name_lower for bad_word in blacklist)
                            
                            if not is_excluded:
                                file_name = f"musinsa_{label_name}_{self.download_counts[label_name] + 1}.jpg"
                                status = self._download_image(img_url, save_path, file_name)
                                
                                if status == "SUCCESS":
                                    self.download_counts[label_name] += 1
                                    page_success += 1
                                    print(f"[저장] {product_name}")
                                elif status == "SKIP":
                                    page_skipped += 1
                                    
                    print(f"✅ {page}페이지 완료 (저장: {page_success}장 | 스킵: {page_skipped}장 | 누적: {self.download_counts[label_name]}장)")
                    
                    # 💡 [핵심 강화 로직] 스킵이 몇 개든 상관없이, '새로 저장한 코트'가 0개면 무의미한 페이지로 간주!
                    if page_success == 0:
                        empty_page_streak += 1
                        print(f"⚠️ 새로운 코트가 없습니다. (연속 허탕: {empty_page_streak}회)")
                        
                        if empty_page_streak >= 5: # 5회 연속 허탕이면 유령 페이지/중복 구간 함정으로 판단
                            print(f"🚨 유령 페이지/중복 구간 함정 감지! 카테고리 [{cat_id}] 수집을 강제로 마치고 다음으로 넘어갑니다.")
                            break
                    else:
                        empty_page_streak = 0 # 단 한 장이라도 건졌다면 허탕 카운터 초기화
                    
                    page += 1 
                    time.sleep(1) # 서버 과부하 방지
                    
                except Exception as e:
                    print(f"⚠️ {page}페이지 에러 발생, 다음으로 넘어갑니다.")
                    page += 1
                    continue

# ==========================================
# 3. 실행부
# ==========================================
if __name__ == "__main__":
    BASE_DIR = r"C:\Users\user\Desktop\졸작 데이터셋"
    
    crawler = MusinsaCrawler(BASE_DIR)
    
    # 언제든 1페이지부터 돌려도 완벽하게 스킵 및 이어받기가 작동합니다.
    crawler.run_crawling(start_page=1)
    
    crawler.close()
    print("\n🎉 모든 코트 데이터 수집 작업이 성공적으로 완료되었습니다!")
```
